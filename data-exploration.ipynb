```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ax = plt.axes()
sns.heatmap(books.isna().transpose(), cbar=False, ax=ax)

plt.xlabel("Columns")
plt.ylabel("Missing values")

plt.show()
```

```python
import numpy as np
```

```python
books ["missing_description"] = np.where(books['description'].isna(), 1, 0)
books ["age_of_book"] = 2024 - (books["published_year"])
```

```python
column_of_interest = ["num_pages", "missing_description", "age_of_book", "average_rating"]
correlation_matrix = books[column_of_interest].corr(method="spearman")

sns.set_theme(style="white")
plt.figure(figsize=(8,6))
heatmap = sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="Spectral", cbar_kws={"label":"Spearman correlation"})

heatmap.set_title("Correlation matrix")
plt.show()
```

```python
book_missing = books[~(books["description"].isna()) |
       ~(books["num_pages"].isna()) |
       ~(books["average_rating"].isna())  |
       ~(books["published_year"].isna())].copy()
```

```python
book_missing
```

```python
book_missing["categories"].value_counts().reset_index().sort_values("count", ascending=False)
```

```python
book_missing["words_in_description"] = books["description"].str.split().str.len()
```

```python
books
```

```python
book_missing.loc[book_missing["words_in_description"].between(25, 34), "description"]
```

```python
book_missing_25_words = book_missing[book_missing["words_in_description"] >= 25].copy()
```

```python
book_missing_25_words
```

```python
book_missing_25_words["title_and_subtitle"] = (
    np.where(book_missing_25_words["subtitle"].isna(), book_missing_25_words["title"],
             book_missing_25_words[["title", "subtitle"]].astype(str).agg(": ".join, axis=1))
)
```

```python
book_missing_25_words
```

```python
book_missing_25_words["tagged_description"] = book_missing_25_words[["isbn13", "description"]].astype(str).agg(" ".join, axis=1)
```

```python
book_missing_25_words
```

```python
(book_missing_25_words
.drop(["subtitle", "missing_description", "age_of_book", "words_in_description"], axis=1)
.to_csv("books_cleaned.csv", index=False)
)
```
